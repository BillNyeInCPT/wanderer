```python
#Imports
from wanderer import *
import numpy as np
import matplotlib.pyplot as plt

#Check stochastic integral over square brownian motion

#create memory for the stochastic integral of the squared brownian motion 
integral_memory = 1000
stochastic_integral = np.zeros(integral_memory)
seed = 123
#loop through paths and calculate the stochastic integral of the squared brownian motion using the empirical_stochastic_integral function
for i in range(1000):
    # generate a brownian motion path
    brownian_motion_generator = BrownianMotionGenerator(mu = 0, sigma = 1, time_step = 0.01, time_horizon = 10, num_paths = 1, seed = 123+i)
    brownian_motion_process = brownian_motion_generator.generate_BrownianMotion()
    #square the path values and create new stochastic process object
    squared_brownian_motion_paths = brownian_motion_process.paths ** 2
    squared_brownian_motion_process = StochasticProcess(paths=squared_brownian_motion_paths, timeHorizon=brownian_motion_process.timeHorizon, timeStep=brownian_motion_process.timeStep, num_paths=brownian_motion_process.num_paths, method='Squared Brownian Motion')
    stochastic_integral[i] = empirical_stochastic_integral(squared_brownian_motion_process.paths[0], brownian_motion_process.paths[0])

#Print the mean and standard deviation of the stochastic integral
print("Mean of stochastic integral of squared brownian motion: ", np.mean(stochastic_integral))
print("Standard deviation of stochastic integral of squared brownian motion: ", np.std(stochastic_integral))
```
